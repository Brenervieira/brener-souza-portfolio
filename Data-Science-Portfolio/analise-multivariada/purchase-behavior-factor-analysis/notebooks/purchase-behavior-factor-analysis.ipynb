# Teste de Barlett 

**VerificaÃ§Ã£o do presuposto de esferacidade, esse teste indica se os dados sÃ£o adequados para anÃ¡lise fatorial**

```python
chi_square_value, p_value = calculate_bartlett_sphericity(df)
```

```python
print('EstatÃ­stica do Teste:' , chi_square_value, '\nValor-p ', p_value)
```

 HipÃ³teses do teste:

H0 (hipÃ³tese nula): A matriz de correlaÃ§Ã£o Ã© uma matriz identidade.
â†’ Isso significa que as variÃ¡veis nÃ£o estÃ£o correlacionadas entre si, logo nÃ£o faria sentido aplicar anÃ¡lise fatorial.

H1 (hipÃ³tese alternativa): A matriz de correlaÃ§Ã£o nÃ£o Ã© identidade.
â†’ Ou seja, existem correlaÃ§Ãµes significativas entre as variÃ¡veis, justificando a anÃ¡lise fatorial.

    

**Quando o resultado retorna p = 0 (na prÃ¡tica, Ã© um valor tÃ£o pequeno que o software arredonda para zero), isso significa que a probabilidade de a matriz ser identidade Ã© praticamente nula. Portanto, rejeitamos fortemente H0 e concluÃ­mos que existe correlaÃ§Ã£o suficiente entre as variÃ¡veis.**

# Analise Fatorial

```python
fa = FactorAnalyzer(rotation= "varimax")
```

```python
# Treianr o Modelo
fa.fit(df)
```

```python
# Obtendo os eigenvalues
ev, v = fa.get_eigenvalues()

# GrÃ¡fico de cotovelo (Scree Plot)
plt.scatter(range(1, df.shape[1]+1), ev)
plt.plot(range(1, df.shape[1]+1), ev)
plt.title('GrÃ¡fico de Cotovelo (Scree Plot)')
plt.xlabel('Fatores')
plt.ylabel('Eigenvalues')
plt.grid(True)
plt.show()
```

```python
fa = FactorAnalyzer(n_factors=2, rotation="varimax")
fa.fit(df)
```

```python
loadings = fa.loadings_
print(loadings)
```

```python
fa.get_communalities()
```
