### DecisÃ£o sobre a variÃ¡vel `income_composition_of_resources`

A anÃ¡lise do fator de inflaÃ§Ã£o da variÃ¢ncia identificou valores de VIF mais elevados para as variÃ¡veis `income_composition_of_resources` e `schooling`, indicando a presenÃ§a de multicolinearidade entre algumas das variÃ¡veis preditoras.

Entretanto, o VIF nÃ£o deve ser utilizado isoladamente como critÃ©rio automÃ¡tico para exclusÃ£o de variÃ¡veis. Embora a multicolinearidade possa afetar a estabilidade e a interpretaÃ§Ã£o individual dos coeficientes de uma regressÃ£o linear, a remoÃ§Ã£o de uma variÃ¡vel tambÃ©m pode provocar perda de informaÃ§Ã£o preditiva relevante.

Para avaliar esse efeito, foram comparadas duas versÃµes do modelo de regressÃ£o linear: uma mantendo a variÃ¡vel `income_composition_of_resources` e outra excluindo-a.

O modelo com a variÃ¡vel apresentou RÂ² de teste de aproximadamente 0,794, MAE de 2,94 e RMSE de 4,09. ApÃ³s a exclusÃ£o, o RÂ² de teste caiu para aproximadamente 0,727, enquanto o MAE aumentou para 3,42 e o RMSE para 4,70.

Os resultados demonstram que a retirada de `income_composition_of_resources` reduziu a capacidade preditiva do modelo. Dessa forma, optou-se por manter a variÃ¡vel na modelagem, considerando que seu ganho de desempenho foi relevante e que o VIF elevado, por si sÃ³, nÃ£o justifica sua exclusÃ£o. A possÃ­vel influÃªncia da multicolinearidade serÃ¡ posteriormente avaliada por meio de modelos regularizados, como Ridge, Lasso e Elastic Net.


| CenÃ¡rio                               | RÂ² teste |  MAE | RMSE |
| ------------------------------------- | -------: | ---: | ---: |
| Com `income_composition_of_resources` |    0,794 | 2,94 | 4,09 |
| Sem `income_composition_of_resources` |    0,727 | 3,42 | 4,70 |

Como a exclusÃ£o da variÃ¡vel reduziu o RÂ² e aumentou os erros de previsÃ£o, a versÃ£o com income_composition_of_resources foi selecionada como baseline definitivo.

# VersÃ£o do Modelo 2 - RegressÃ£o Lasso

```python
from sklearn.linear_model import Lasso
from sklearn import metrics
import numpy as np

# Modelo V2 - RegressÃ£o Lasso
modelo_v2 = Lasso(
    alpha=1.0,
    max_iter=10000
)

# Treinamento
modelo_v2.fit(X_train_scaled, y_train)

# PrevisÃµes
y_pred_train_v2 = modelo_v2.predict(X_train_scaled)
y_pred_test_v2 = modelo_v2.predict(X_test_scaled)

# MÃ©tricas de treino
mae_train_v2 = metrics.mean_absolute_error(
    y_train,
    y_pred_train_v2
)

mse_train_v2 = metrics.mean_squared_error(
    y_train,
    y_pred_train_v2
)

rmse_train_v2 = np.sqrt(mse_train_v2)

r2_train_v2 = metrics.r2_score(
    y_train,
    y_pred_train_v2
)

# MÃ©tricas de teste
mae_test_v2 = metrics.mean_absolute_error(
    y_test,
    y_pred_test_v2
)

mse_test_v2 = metrics.mean_squared_error(
    y_test,
    y_pred_test_v2
)

rmse_test_v2 = np.sqrt(mse_test_v2)

r2_test_v2 = metrics.r2_score(
    y_test,
    y_pred_test_v2
)

# Resultados
print("Modelo V2 - RegressÃ£o Lasso | alpha = 1.0")

print("\nResultados de Treino")
print("Mean Absolute Error (MAE) - Treino:", mae_train_v2)
print("Mean Squared Error (MSE) - Treino:", mse_train_v2)
print("Root Mean Squared Error (RMSE) - Treino:", rmse_train_v2)
print("RÂ² Score - Treino:", r2_train_v2)

print("\nResultados de Teste")
print("Mean Absolute Error (MAE) - Teste:", mae_test_v2)
print("Mean Squared Error (MSE) - Teste:", mse_test_v2)
print("Root Mean Squared Error (RMSE) - Teste:", rmse_test_v2)
print("RÂ² Score - Teste:", r2_test_v2)
```
