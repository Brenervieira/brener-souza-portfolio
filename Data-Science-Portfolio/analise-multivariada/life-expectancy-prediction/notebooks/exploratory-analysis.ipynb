## _Outliers_

```python

# VariÃ¡veis com mais de 5% de valores ausentes
variaveis = [
    "population",
    "hepatitis_b",
    "gdp",
    "total_expenditure",
    "alcohol",
    "income_composition_of_resources",
    "schooling"
]

resultado_iqr = []

for coluna in variaveis:

    # Remove os valores ausentes para calcular o IQR
    serie = df_dados[coluna].dropna()

    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    qtd_outliers = (
        (serie < limite_inferior) |
        (serie > limite_superior)
    ).sum()

    percentual = (qtd_outliers / len(serie)) * 100

    resultado_iqr.append({
        "VariÃ¡vel": coluna,
        "Q1": round(q1, 2),
        "Q3": round(q3, 2),
        "IQR": round(iqr, 2),
        "Limite Inferior": round(limite_inferior, 2),
        "Limite Superior": round(limite_superior, 2),
        "Qtd. Outliers": qtd_outliers,
        "% Outliers": round(percentual, 2)
    })

resultado_iqr = pd.DataFrame(resultado_iqr)

resultado_iqr.sort_values("% Outliers", ascending=False)
```

## _RemoÃ§Ã£o de Outiliers_

```python
novo_df_dados = df_dados.copy()

mascara_manter = pd.Series(True, index=df_dados.index)

resumo_remocao = []

for coluna in variaveis:

    serie = df_dados[coluna].dropna()

    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    mascara_outliers = (
        (df_dados[coluna] < limite_inferior) |
        (df_dados[coluna] > limite_superior)
    )

    resumo_remocao.append({
        "variavel": coluna,
        "limite_inferior": round(limite_inferior, 2),
        "limite_superior": round(limite_superior, 2),
        "registros_identificados": mascara_outliers.sum()
    })

    # Remove outliers, mas preserva valores ausentes
    mascara_manter &= (
        ~mascara_outliers |
        df_dados[coluna].isna()
    )

novo_df_dados = (
    df_dados[mascara_manter]
    .copy()
    .reset_index(drop=True)
)

resumo_remocao = pd.DataFrame(resumo_remocao)

print(f"Quantidade inicial: {len(df_dados)}")
print(f"Quantidade final: {len(novo_df_dados)}")
print(f"Total removido: {len(df_dados) - len(novo_df_dados)}")
print(
    f"Percentual removido: "
    f"{(1 - len(novo_df_dados) / len(df_dados)) * 100:.2f}%"
)

resumo_remocao
```

## _ImputaÃ§Ã£o da Mediana_

```python
def preencher_ausentes_mediana(dataframe):
    """
    Preenche os valores ausentes das colunas numÃ©ricas com a mediana.

    Retorna:
    - DataFrame tratado
    - Resumo da imputaÃ§Ã£o
    """

    df_tratado = dataframe.copy()
    resumo = []

    colunas_numericas = df_tratado.select_dtypes(
        include=np.number
    ).columns

    for coluna in colunas_numericas:
        quantidade_ausentes = df_tratado[coluna].isna().sum()

        if quantidade_ausentes > 0:
            mediana = df_tratado[coluna].median()

            df_tratado[coluna] = (
                df_tratado[coluna].fillna(mediana)
            )

            resumo.append({
                "variavel": coluna,
                "mediana_utilizada": round(mediana, 2),
                "valores_preenchidos": quantidade_ausentes
            })

    resumo_imputacao = pd.DataFrame(resumo)

    return df_tratado, resumo_imputacao
```

```python
def filtrar_e_visualizar_correlacao(
    dataframe,
    limite=0.30,
    remover_coluna=None
):
    """
    Calcula a matriz de correlaÃ§Ã£o, filtra os valores pelo limite
    definido e exibe somente as correlaÃ§Ãµes relevantes.

    ParÃ¢metros
    ----------
    dataframe : pandas.DataFrame
        DataFrame que serÃ¡ analisado.

    limite : float
        Valor mÃ­nimo de correlaÃ§Ã£o absoluta que serÃ¡ exibido.

    remover_coluna : str ou list, opcional
        Coluna ou lista de colunas que nÃ£o devem aparecer no grÃ¡fico.

    Retorna
    -------
    pandas.DataFrame
        Matriz de correlaÃ§Ã£o filtrada.
    """

    # Seleciona somente variÃ¡veis numÃ©ricas
    dados_numericos = dataframe.select_dtypes(
        include=np.number
    )

    # Calcula a matriz de correlaÃ§Ã£o
    correlacao = dados_numericos.corr()

    # MantÃ©m correlaÃ§Ãµes acima do limite e remove a diagonal principal
    filtro = (
        (correlacao.abs() >= limite)
        & (correlacao != 1.0)
    )

    # Remove linhas e colunas completamente vazias
    correlacao_filtrada = (
        correlacao
        .where(filtro)
        .dropna(how="all")
        .dropna(axis=1, how="all")
    )

    # Remove colunas especÃ­ficas, quando solicitado
    if remover_coluna is not None:

        if isinstance(remover_coluna, str):
            remover_coluna = [remover_coluna]

        correlacao_filtrada = correlacao_filtrada.drop(
            index=remover_coluna,
            columns=remover_coluna,
            errors="ignore"
        )

    # Exibe o grÃ¡fico
    plt.figure(figsize=(16, 10))

    sns.heatmap(
        correlacao_filtrada,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        linewidths=0.5
    )

    plt.title(
        f"CorrelaÃ§Ãµes com valor absoluto â‰¥ {limite}"
    )
    plt.tight_layout()
    plt.show()

    return correlacao_filtrada
```

```python
matriz_filtrada = filtrar_e_visualizar_correlacao(
    dataframe=novo_df_dados,
    limite=0.30
)
```

```python
matriz_sem_alvo = filtrar_e_visualizar_correlacao(
    dataframe=novo_df_dados,
    limite=0.55,
    remover_coluna="life_expectancy"
)

correlacoes_em_colunas = (
    matriz_sem_alvo
    .stack()
    .reset_index()
)

correlacoes_em_colunas.columns = [
    "variavel_1",
    "variavel_2",
    "correlacao"
]

# Remove pares duplicados: A-B e B-A
correlacoes_em_colunas["par"] = correlacoes_em_colunas.apply(
    lambda linha: tuple(
        sorted([linha["variavel_1"], linha["variavel_2"]])
    ),
    axis=1
)

correlacoes_em_colunas = (
    correlacoes_em_colunas
    .drop_duplicates(subset="par")
    .drop(columns="par")
    .assign(
        correlacao_absoluta=lambda df: df["correlacao"].abs()
    )
    .sort_values(
        by="correlacao_absoluta",
        ascending=False
    )
    .reset_index(drop=True)
)

correlacoes_em_colunas
```
