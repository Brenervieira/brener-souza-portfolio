## Verificando a porcentagem da distribuição da variavel alvo

```python
# Comvertendo a coluna Target para int
def prevalencia(y_actual):
    return sum(y_actual) / len(y_actual)


print(f"Porcentagem da classe positiva no df: {prevalencia(df['LABEL_TARGET'].values)} %" )
```

*Como a porcentagem de aceites no DataFrame é relativamente pequena, utilizaremos a amostragem aleatória para selecionar uma amostra representativa. O objetivo é preservar as características originais do conjunto de dados, mas organizá-lo em três subconjuntos distintos: treino, teste e validação.*

### Divisão do Dataset e Amostragem

```python
df_data = df.sample(n = len(df))
df_data = df_data.reset_index(drop=True)
```

```python
# Dvisão do Dataset em Treino , validação e teste

amostra_30 = df_data.sample(frac=0.3)

df_valid = amostra_30.sample(frac=0.5)

df_test = amostra_30.drop(df_valid.index)

df_treino = df_data.drop(amostra_30.index)

```

```python
# Verificando Distribuição das amostras

# Aplicando nos conjuntos
print(f"Treino: {prevalencia(df_treino['LABEL_TARGET'].values) * 100:.2f}%")
print(f"Validação: {prevalencia(df_valid['LABEL_TARGET'].values) * 100:.2f}%")
print(f"Teste: {prevalencia(df_test['LABEL_TARGET'].values) * 100:.2f}%")
```

# Balanceameto de Classe

```python
df_treino.shape
```

```python
df_valid.shape
```

```python
df_test.shape
```

```python
df_test['LABEL_TARGET'].value_counts()
```

```python
def undersample(df, target_col="LABEL_TARGET", random_state=42):
    # separa classes
    classe_majoritaria = df[df[target_col] == 0]
    classe_minoritaria = df[df[target_col] == 1]

    # pega o tamanho da classe minoritária
    n_minoritaria = len(classe_minoritaria)

    # reduz a classe majoritária para o mesmo tamanho
    classe_majoritaria_reduzida = classe_majoritaria.sample(
        n=n_minoritaria, random_state=random_state
    )

    # junta novamente
    df_balanceado = pd.concat([classe_majoritaria_reduzida, classe_minoritaria])

    # embaralha
    df_balanceado = df_balanceado.sample(frac=1, random_state=random_state).reset_index(drop=True)

    return df_balanceado

# aplicando no conjunto
df_treino_bal = undersample(df_treino, "LABEL_TARGET")


# conferindo distribuição
print("Treino balanceado:\n", df_treino_bal['LABEL_TARGET'].value_counts())
```

# Verificando a Distribuição do Conjunto  e Salvando o Pré-Processamento



```python
# Verificando Distribuição das amostras
print(f"Balanceamento em Treino: {prevalencia(df_treino_bal['LABEL_TARGET'].values) * 100:.2f}%")

```

```python
# Salavando os dados

df_treino.to_csv('dados/dados_treino.csv',index=False)
df_test.to_csv('dados/dados_teste.csv',index=False)
df_treino_bal.to_csv('dados/dados_treino_final.csv',index=False)
df_valid.to_csv('dados/dados_valid.csv',index=False)
```
