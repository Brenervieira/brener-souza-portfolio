# Carregando Dados e Limpeza Inicial

```python
df = pd.read_csv('../data/dataset.csv')
```

```python
# Padronizando nome de  Colunas

df.columns = (
    df.columns
    .str.strip()        # remove espaÃ§os no inÃ­cio e no fim
    .str.lower()        # transforma em minÃºsculas
    .str.replace(" ", "_", regex=False)  # troca espaÃ§os internos por _
)
```

```python
# Verificando nulos

df.isnull().sum()
```

```python
# Tipos de Dados

df.info()
```

```python
# Excluir a coluna ID
df = df.drop(columns=["id"])

# Excluir somente os registros com renda ausente
df = df.dropna(subset=["renda"])

# Reorganizar o Ã­ndice
df = df.reset_index(drop=True)

print("DimensÃ£o final:", df.shape)
print("Valores ausentes em renda:", df["renda"].isna().sum())
print("Colunas:", df.columns.tolist())
```

A coluna `id` foi excluÃ­da por funcionar apenas como identificador dos clientes, nÃ£o contribuindo para a identificaÃ§Ã£o dos fatores latentes. TambÃ©m foram removidos os 24 registros com valor ausente na variÃ¡vel `renda`, correspondentes a aproximadamente 1,07% da base. Como essa proporÃ§Ã£o Ã© pequena, a exclusÃ£o nÃ£o compromete significativamente o conjunto de dados.


```python
df.head(3)
```

```python
# ConversÃ£o de data em cadastro cliente

df["data_cadastro_cliente"] = pd.to_datetime(
    df["data_cadastro_cliente"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
# Definir o ano atual
ano_atual = 2026

# Criar o tempo de relacionamento do cliente
df["anos_cliente"] = (
    ano_atual - df["data_cadastro_cliente"].dt.year
)

# Criar a idade do cliente
df["idade_cliente"] = (
    ano_atual - df["ano_nascimento"]
)
df = df.drop(
    columns=[
        "data_cadastro_cliente",
        "data_cadastro_formatada",
        "ano_nascimento"
    ],
    errors="ignore"
)
```

As variÃ¡veis `ano_nascimento` e `data_cadastro_cliente` foram convertidas, respectivamente, em `idade_cliente` e `anos_cliente`, tornando as informaÃ§Ãµes mais adequadas para a anÃ¡lise fatorial. ApÃ³s a criaÃ§Ã£o dessas novas variÃ¡veis numÃ©ricas, as colunas originais foram excluÃ­das por nÃ£o acrescentarem informaÃ§Ã£o relevante ao modelo.


```python
print(df['educacao'].value_counts())
print('\n', df['estado_civil'].value_counts())
```

```python


# Padronizar os textos
df["educacao"] = (
    df["educacao"]
    .str.strip()
    .str.lower()
)

df["estado_civil"] = (
    df["estado_civil"]
    .str.strip()
    .str.lower()
)

mapa_educacao = {
    "ensino fundamental": 1,
    "ensino mÃ©dio": 2,
    "graduaÃ§Ã£o": 3,
    "mestrado": 4,
    "phd": 5
}

df["nivel_educacao"] = df["educacao"].map(mapa_educacao)

print(df[["educacao", "nivel_educacao"]].value_counts())
print("Valores nÃ£o convertidos:", df["nivel_educacao"].isna().sum())
```

```python
df["estado_civil"] = pd.Categorical(
    df["estado_civil"],
    categories=[
        "casado",
        "solteiro",
        "divorciado",
        "viÃºvo"
    ]
)

dummies_estado_civil = pd.get_dummies(
    df["estado_civil"],
    prefix="estado_civil",
    drop_first=True,
    dtype=int
)

df = pd.concat(
    [df, dummies_estado_civil],
    axis=1
)

df = df.drop(
    columns=[
        "educacao",
        "estado_civil"
    ]
)
```

```python
df[
    [
        "nivel_educacao",
        "estado_civil_solteiro",
        "estado_civil_divorciado",
        "estado_civil_viÃºvo"
    ]
].head()
```

As variÃ¡veis `educacao` e `estado_civil` foram transformadas em formato numÃ©rico para permitir sua inclusÃ£o na anÃ¡lise fatorial. A escolaridade foi convertida em uma escala ordinal, respeitando a progressÃ£o entre os nÃ­veis de ensino. JÃ¡ o estado civil foi representado por variÃ¡veis indicadoras, evitando a criaÃ§Ã£o de uma ordem artificial entre categorias que nÃ£o possuem hierarquia natural.


```python
df["renda"] = pd.to_numeric(
    df["renda"]
    .astype("string")
    .str.strip()
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False),
    errors="coerce"
)
```

A variÃ¡vel `renda` foi regularizada por meio da remoÃ§Ã£o do sÃ­mbolo monetÃ¡rio, dos separadores de milhar e dos espaÃ§os excedentes, permitindo sua conversÃ£o de texto para formato numÃ©rico. Em seguida, as variÃ¡veis selecionadas foram padronizadas, colocando-as em uma escala comum, com mÃ©dia zero e desvio-padrÃ£o igual a um. Esse procedimento reduz a influÃªncia das diferenÃ§as de magnitude entre renda, gastos, idade, tempo de relacionamento e demais variÃ¡veis na formaÃ§Ã£o dos fatores.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

dados_padronizados = pd.DataFrame(
    scaler.fit_transform(df),
    columns=df.columns,
    index=df.index
)
```

```python
print('KMO Global:' , kmo_model)
```

```python
print('KMO por Variavel:' , kmo_all)
```

```python
kmo_variaveis = pd.DataFrame({
    "variavel": dados_padronizados.columns,
    "kmo": kmo_all
}).sort_values("kmo")

display(kmo_variaveis.round(3))
```

```python
colunas_estado_civil = [
    "estado_civil_divorciado",
    "estado_civil_solteiro",
    "estado_civil_viÃºvo"
]

df = df.drop(
    columns=colunas_estado_civil,
    errors="ignore"
)
```

O teste de esfericidade de Bartlett apresentou resultado significativo, com estatÃ­stica de 19.739 e valor-p prÃ³ximo de zero, indicando a existÃªncia de correlaÃ§Ãµes suficientes entre as variÃ¡veis para a aplicaÃ§Ã£o da anÃ¡lise fatorial. O KMO global foi de 0,87, demonstrando boa adequaÃ§Ã£o da amostra. Entretanto, as variÃ¡veis indicadoras de estado civil apresentaram baixos valores individuais de KMO, especialmente `estado_civil_divorciado`, com 0,465. Por esse motivo, as categorias de estado civil foram removidas da anÃ¡lise fatorial principal, podendo ser utilizadas posteriormente para caracterizar os clientes a partir dos fatores identificados.


```python
from sklearn.preprocessing import StandardScaler
from factor_analyzer.factor_analyzer import calculate_kmo

scaler = StandardScaler()

dados_padronizados = pd.DataFrame(
    scaler.fit_transform(df),
    columns=df.columns,
    index=df.index
)

kmo_all_novo, kmo_global_novo = calculate_kmo(dados_padronizados)

print(f"Novo KMO global: {kmo_global_novo:.3f}")

tabela_kmo_nova = pd.DataFrame({
    "variavel": dados_padronizados.columns,
    "kmo": kmo_all_novo
}).sort_values("kmo")

display(tabela_kmo_nova.round(3))
```

ApÃ³s a remoÃ§Ã£o das variÃ¡veis de estado civil, o KMO global aumentou de 0,87 para 0,881, indicando melhora na adequaÃ§Ã£o da base Ã  anÃ¡lise fatorial. AlÃ©m disso, todas as variÃ¡veis restantes apresentaram KMO individual acima de 0,50, valor mÃ­nimo geralmente aceito. Dessa forma, considera-se que o conjunto de dados apresenta condiÃ§Ãµes adequadas para o prosseguimento da anÃ¡lise fatorial, iniciando pela avaliaÃ§Ã£o dos autovalores e do grÃ¡fico de sedimentaÃ§Ã£o para definiÃ§Ã£o do nÃºmero de fatores.


# AnÃ¡lise Fatorial - Varimax 

```python
# Modelo exploratÃ³rio inicial
fa_inicial = FactorAnalyzer(rotation="varimax")
fa_inicial.fit(dados_padronizados)
```

```python
from factor_analyzer import FactorAnalyzer
import matplotlib.pyplot as plt
import pandas as pd

# Quantidade total de variÃ¡veis
n_variaveis = dados_padronizados.shape[1]

# Modelo exploratÃ³rio inicial, sem rotaÃ§Ã£o
fa_inicial = FactorAnalyzer(
    n_factors=n_variaveis,
    method="minres",
    rotation=None
)

fa_inicial.fit(dados_padronizados)

# Obter os autovalores
autovalores, autovalores_fatores = fa_inicial.get_eigenvalues()

numero_fatores = range(1, len(autovalores) + 1)

plt.figure(figsize=(10, 6))

plt.plot(
    numero_fatores,
    autovalores,
    marker="o",
    linewidth=2
)

# Linha de referÃªncia do critÃ©rio de Kaiser
plt.axhline(
    y=1,
    linestyle="--",
    label="Autovalor = 1"
)

plt.title("GrÃ¡fico de sedimentaÃ§Ã£o â€” Scree Plot")
plt.xlabel("NÃºmero do fator")
plt.ylabel("Autovalor")
plt.xticks(numero_fatores)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Modelo exploratÃ³rio inicial
fa_inicial_v2 = FactorAnalyzer(n_factors= 4, rotation="varimax")
fa_inicial_v2.fit(dados_padronizados)
```

```python
loadings_varimax = pd.DataFrame(
    fa_inicial_v2.loadings_,
    index=dados_padronizados.columns,
    columns=[
        "Fator 1",
        "Fator 2",
        "Fator 3",
        "Fator 4"
    ]
)

display(loadings_varimax.round(3))
```

```python
fa_inicial_v2.get_communalities()
```

```python


# Modelo com 4 fatores e rotaÃ§Ã£o oblÃ­qua
fa_inicial_v2_obliquo = FactorAnalyzer(
    n_factors=4,
    method="minres",
    rotation="oblimin"
)

fa_inicial_v2_obliquo.fit(dados_padronizados)
```

```python
cargas_oblimin = pd.DataFrame(
    fa_inicial_v2_obliquo.loadings_,
    index=dados_padronizados.columns,
    columns=[
        "Fator 1",
        "Fator 2",
        "Fator 3",
        "Fator 4"
    ]
)

display(cargas_oblimin.round(3))
```

```python
nomes_fatores = ["Fator 1", "Fator 2", "Fator 3", "Fator 4"]

correlacao_fatores = pd.DataFrame(
    fa_inicial_v2_obliquo.phi_,
    index=nomes_fatores,
    columns=nomes_fatores
)

display(correlacao_fatores.round(3))
```

A matriz de correlaÃ§Ã£o dos fatores revelou associaÃ§Ãµes moderadas entre algumas dimensÃµes, destacando-se a correlaÃ§Ã£o positiva de 0,453 entre os fatores 1 e 3 e a correlaÃ§Ã£o negativa de âˆ’0,371 entre os fatores 1 e 4. Como esses valores ultrapassam 0,30 em magnitude, conclui-se que os fatores nÃ£o sÃ£o totalmente independentes. Dessa forma, optou-se pela rotaÃ§Ã£o oblÃ­qua, que permite correlaÃ§Ã£o entre os fatores e representa de maneira mais realista as relaÃ§Ãµes existentes entre consumo, perfil familiar, campanhas e comportamento digital dos clientes.


A anÃ¡lise fatorial com rotaÃ§Ã£o oblÃ­qua identificou quatro dimensÃµes principais associadas ao comportamento dos clientes.

**Fator 1 â€” Intensidade e diversidade de consumo:** reÃºne os gastos com peixes, frutas, doces, carnes e outros produtos, alÃ©m das compras realizadas pela web, catÃ¡logo e loja fÃ­sica. Esse fator representa clientes com maior volume e diversidade de consumo, que utilizam diferentes canais de compra.

**Fator 2 â€” Receptividade Ã s campanhas de marketing:** concentra as variÃ¡veis relacionadas Ã  aceitaÃ§Ã£o das campanhas promocionais, especialmente as campanhas 2, 3, 4, 5 e 6, alÃ©m do gasto com vinhos. Esse fator caracteriza clientes mais responsivos Ã s aÃ§Ãµes de marketing e com maior afinidade com essa categoria de produto.

**Fator 3 â€” Perfil familiar e etapa de vida:** Ã© composto principalmente pela presenÃ§a de adolescentes em casa, idade do cliente e utilizaÃ§Ã£o dos canais web e loja. Esse fator indica que a composiÃ§Ã£o familiar e a fase de vida do cliente influenciam suas escolhas de compra e sua utilizaÃ§Ã£o dos diferentes canais.

**Fator 4 â€” Comportamento digital e sensibilidade a promoÃ§Ãµes:** apresenta maior associaÃ§Ã£o com o nÃºmero de visitas mensais ao site e com as compras promocionais. A relaÃ§Ã£o negativa com a renda sugere que clientes de menor poder aquisitivo tendem a visitar mais o ambiente digital e buscar oportunidades promocionais.

Conclui-se que o comportamento de consumo dos clientes Ã© explicado principalmente pela intensidade e diversidade dos gastos, pela receptividade Ã s campanhas de marketing, pelo perfil familiar e pela interaÃ§Ã£o com os canais digitais e promocionais. Esses fatores podem apoiar a segmentaÃ§Ã£o da base e a criaÃ§Ã£o de estratÃ©gias comerciais mais direcionadas Ã s caracterÃ­sticas de cada grupo de clientes.
